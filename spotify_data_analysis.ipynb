# Data Cleaning and Preprocessing

### Check for Missing Values

These code checks both datasets for missing values in each column to determine whether cleaning is necessary before merging.

```python
df_modern.isnull().sum()
```

```python
df_classic.isnull().sum()
```

### Handling Missing Values

A few of the missing values were found in artist_popularity, artist_followers, artist_genres, track_name, artists_genres, and album_name. Because the number of missing observations was very small in comparison to the total dataset size, these rows were removed. Dropping these rows prevents inaccurate statistical calculations while preserving the overall dataset integrity.

```python
df_modern = df_modern.dropna(subset=["artist_name", "track_name"])
df_classic = df_classic.dropna(subset=["artist_name", "track_name", 'artist_popularity', 'artist_followers', 'artist_genres', 'album_name'])
```

However, for df_modern, there were a significant amount of missing values in artists_genres. Thus, dropping these rows would have cause a significant loss in data, leading to innacurate findings. Rather than dropping them, I decided to fill these missing values with "Unknown" in order to preserve the data.

```python
df_modern["artist_genres"] = (df_modern["artist_genres"].replace("[]", np.nan).fillna("Unknown"))
```

### Verifying Missing Values Were Removed

```python
df_modern.isnull().sum()
```

```python
df_classic.isnull().sum()
```

### Convert Duration Column to Ensure Consistency

The track_duration_ms column in the classic dataset was originally stored in milliseconds, while the modern dataset used minutes (track_duration_min). To ensure consistency before merging, the duration values in the classic dataset were converted from milliseconds to minutes. After conversion, the original millisecond column was removed to avoid redundancy.

```python
df_classic["track_duration_min"] = df_classic["track_duration_ms"] / 60000
```

```python
df_classic = df_classic.drop(columns=["track_duration_ms"])
```

### Extract Release Year from Album Release Date

The album_release_date column contained inconsistent formats (some full dates and some year-only values). Instead of converting the column to full datetime format, the first four characters were extracted to geet the release year. This ensures consistent numerical year values across both datasets.

```python
df_modern["release_year"] = df_modern["album_release_date"].astype(str).str[:4].astype(int)
df_classic["release_year"] = df_classic["album_release_date"].astype(str).str[:4].astype(int)
```

### Drop the Original Release Date Column

Since the analysis only requires the release year, the original album_release_date column is dropped to simplify the dataset and avoid duplications and repetitiveness.

```python
df_modern = df_modern.drop(columns=["album_release_date"])
df_classic = df_classic.drop(columns=["album_release_date"])
```

### Create Release Decade Feature

Grouped the data by decade to simplify analysis of trends.

```python
df_modern["release_decade"] = (df_modern["release_year"] // 10) * 10
df_classic["release_decade"] = (df_classic["release_year"] // 10) * 10
```

### Create Duration Category Feature

To simplify analysis, track duration was grouped into three categories based on length (Short, Medium, and Long). This feature helps analyze whether song length influences popularity.

```python
df_modern["duration_category"] = pd.cut(
    df_modern["track_duration_min"],
    bins=[0, 2.5, 4, df_modern["track_duration_min"].max()],
    labels=["Short", "Medium", "Long"])

df_classic["duration_category"] = pd.cut(
    df_classic["track_duration_min"],
    bins=[0, 2.5, 4, df_classic["track_duration_min"].max()],
    labels=["Short", "Medium", "Long"])
```

### Merge the Modern and Classic Datasets

Since both datasets now share identical column names and consistent units, they were merged vertically into a single combined dataset for analysis.

```python
combined_df = pd.concat([df_modern, df_classic], ignore_index=True)
```

### Verify Merge 

This confirms the size of the combined dataset after merging.

```python
combined_df.shape
```

### Create Popularity Level Category Feature

In order to simplify analysis, track popularity was grouped into three categories: Low, Medium, and High. This allows for easier comparison across groups. 

```python
combined_df["popularity_level"] = pd.cut(
    combined_df["track_popularity"],
    bins=[0, 40, 70, 100],
    labels=["Low", "Medium", "High"]
)
```

### Convert Selected Variables to Categorical Type

Certain variables were converted to categorical type to improve grouping efficiency and visualization clarity.

```python
combined_df["explicit"] = combined_df["explicit"].astype("category")
combined_df["album_type"] = combined_df["album_type"].astype("category")
combined_df["popularity_level"] = combined_df["popularity_level"].astype("category")
combined_df["duration_category"] = combined_df["duration_category"].astype("category")
```

# Data Visualization

## Research Question 1 

### Do explicit songs have higher average track popularity than non-explicit songs?

### Histogram of Track Popularity

This histogram shows the overall distribution of track popularity scores. It helps identify the shape, center, and spread of the popularity variable.

```python
plt.hist(combined_df["track_popularity"], bins=20)

plt.title("Distribution of Track Popularity")
plt.xlabel("Track Popularity")
plt.ylabel("Frequency")
plt.show()
```

### Explanation

The histogram shows that track popularity is moderately left-skewed, meaning that most songs have mid to high popularity scores (around 50-80). The center of the distribution appears around 60–70.There is also a noticeable spike at 0 popularity, implying there is a group of songs with very low or no measurable popularity. This may represent newly released tracks or songs with minimal streaming activity. The overall spread covers a range from 0-100 on the track popularity scale.

### Average Popularity by Explicit Status

This code calculates the average track popularity for explicit and non-explicit songs

```python
explicit_means = combined_df.groupby("explicit", observed=True)["track_popularity"].mean()
explicit_means
```

### Bar Char Showing Explicit vs Non-Explicit Popularity

This bar chart visualizes the average popularity for explicit and non-explicit tracks.

```python
explicit_means.plot(kind="bar")

plt.title("Average Track Popularity by Explicit Status")
plt.xlabel("Explicit")
plt.ylabel("Average Track Popularity")
plt.xticks(rotation=0)
plt.show()
```

### Explanation:

The bar chart shows that explicit songs have a higher average track popularity than non-explicit songs. On average, explicit tracks score approximately 7–8 popularity points higher. This suggests that explicit content may be associated with slightly greater listener engagement and appeal.

### Boxplot Showing Popularity by Explicit Status

This boxplot compares the distribution, median, and spread of popularity for explicit and non-explicit songs.

```python
sns.boxplot(x="explicit", y="track_popularity", data=combined_df)

plt.title("Popularity Distribution by Explicit Status")
plt.xlabel("Explicit")
plt.ylabel("Track Popularity")
plt.show()
```

### Explanation:

The boxplot shows that explicit songs have a higher median popularity than non-explicit songs. The interquartile range for explicit tracks is shifted upward, indicating that the middle 50% of explicit songs tend to have higher popularity scores. Both groups show variability across the full 0–100 range, but explicit songs appear to have more low-popularity outliers. Overall, the distribution supports the earlier finding that explicit tracks tend to have higher average popularity.

## Research Question 2

### Does song duration influence popularity?

### Popularity by Duration Category

This code calculates the average popularity for each duration category.

```python
duration_means = combined_df.groupby("duration_category", observed=True)["track_popularity"].mean()
duration_means
```

### Bar Chart: Average Popularity by Duration Category

This bar chart compares the average track popularity across Short, Medium, and Long duration categories to determine whether song length influences popularity.

```python
duration_means.plot(kind="bar")

plt.title("Average Popularity by Duration Category")
plt.xlabel("Duration Category")
plt.ylabel("Average Track Popularity")
plt.xticks(rotation=0)
plt.show()
```

### Explanation: 

The bar chart shows that Short songs have an average popularity of approximately 42, while Medium and Long songs have higher average popularity scores of about 54–55. This indicates that Medium and Long tracks are, on average, about 12–13 points more popular than Short tracks. The similarity between Medium and Long categories suggests that once a song exceeds a short duration, additional length does not significantly increase popularity.

### Boxplot Showing Popularity by Duration Category

This boxplot compares the distribution, median, and variability of track popularity across Short, Medium, and Long duration categories.

```python
sns.boxplot(x="duration_category", y="track_popularity", data=combined_df)

plt.title("Popularity Distribution by Duration Category")
plt.xlabel("Duration Category")
plt.ylabel("Track Popularity")
plt.show()
```

### Explanation

The boxplot shows that Short songs have a median popularity of approximately 45, while Medium and Long songs have higher medians of about 60. This aligns with the bar chart results, where Medium and Long tracks averaged around 54–55, compared to 42 for Short tracks. The interquartile ranges for Medium and Long songs are similar and shifted upward compared to Short songs, indicating that the middle 50% of Medium and Long tracks tend to perform better. Overall, the distribution suggests that very short songs are generally less popular, while Medium and Long songs show comparable popularity levels.

## Research Question 3

### Do artists with more followers tend to have higher track popularity?

### Scatter Plot: Followers vs Popularity

```python
plt.scatter(
    combined_df["artist_followers"],
    combined_df["track_popularity"],
    alpha=0.4
)

plt.xscale("log")

plt.title("Artist Followers vs Track Popularity (Log Scale)")
plt.xlabel("Artist Followers (Log Scale)")
plt.ylabel("Track Popularity")
plt.show()
```

### Explanation:

The scatter plot uses a logarithmic scale for artist followers to account for the wide range of follower counts. The plot shows a positive relationship between artist followers and track popularity, as popularity generally increases with follower size. However, the relationship appears moderate rather than strong, since there is substantial variability in popularity at all follower levels. This suggests that while larger follower bases are associated with higher popularity, follower count alone does not fully determine a song’s success.

### Correlation Calculation

This calculates the correlation coefficient between artist followers and track popularity.

```python
correlation = combined_df["artist_followers"].corr(combined_df["track_popularity"])
correlation
```

### Explanation:

The correlation coefficient between artist followers and track popularity is approximately 0.23, indicating a weak positive relationship. This suggests that artists with more followers tend to have slightly higher track popularity, but follower count alone does not strongly predict a song’s popularity. Although the relationship is positive, the low correlation value indicates that many other factors likely influence track popularity beyond follower count.

## Pair Plot of Key Numerical Variables

```python
sample_df = combined_df[[
    "track_popularity",
    "artist_followers",
    "track_duration_min",
    "explicit"
]].dropna().sample(1000, random_state=42)

sns.pairplot(
    sample_df,
    hue="explicit",
    diag_kind="hist"
)

plt.suptitle("Pair Plot of Popularity, Followers, and Duration", y=1.02)
plt.show()
```

### Explanation:

The pairplot provides a comprehensive view of relationships among track popularity, artist followers, and track duration. The diagonal histograms confirm that popularity is left-skewed with a concentration in the mid-to-high range, while artist followers are highly right-skewed. The scatter plots show a weak positive relationship between followers and popularity, consistent with the correlation coefficient (r ≈ 0.23). In contrast, the relationship between duration and popularity appears weak, with no strong visible pattern. Overall, the pairplot reinforces earlier findings that follower count has some influence on popularity, while duration has a limited effect.
